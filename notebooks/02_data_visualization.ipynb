```python
# notebooks/02_data_visualization.ipynb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Load data
df = pd.read_csv('../data/synthetic_student_data.csv')
print(f"Dataset shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print(f"\nDropout rate: {df['is_dropout'].mean() * 100:.1f}%")
print(f"\nFirst 5 rows:")
df.head()
```

```python
# Dropout by Attendance
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Attendance
df[df['is_dropout']==0]['attendance_rate'].hist(alpha=0.7, label='Non-Dropout', bins=30, ax=axes[0,0], color='green')
df[df['is_dropout']==1]['attendance_rate'].hist(alpha=0.7, label='Dropout', bins=30, ax=axes[0,0], color='red')
axes[0,0].set_title('Attendance Rate by Dropout Status')
axes[0,0].legend()

# Assignments
df[df['is_dropout']==0]['assignments_completed'].hist(alpha=0.7, label='Non-Dropout', bins=30, ax=axes[0,1], color='green')
df[df['is_dropout']==1]['assignments_completed'].hist(alpha=0.7, label='Dropout', bins=30, ax=axes[0,1], color='red')
axes[0,1].set_title('Assignments Completed by Dropout Status')
axes[0,1].legend()

# Test Scores
df[df['is_dropout']==0]['test_scores_avg'].hist(alpha=0.7, label='Non-Dropout', bins=30, ax=axes[1,0], color='green')
df[df['is_dropout']==1]['test_scores_avg'].hist(alpha=0.7, label='Dropout', bins=30, ax=axes[1,0], color='red')
axes[1,0].set_title('Test Scores by Dropout Status')
axes[1,0].legend()

# Days Absent
df[df['is_dropout']==0]['days_absent_last_term'].hist(alpha=0.7, label='Non-Dropout', bins=20, ax=axes[1,1], color='green')
df[df['is_dropout']==1]['days_absent_last_term'].hist(alpha=0.7, label='Dropout', bins=20, ax=axes[1,1], color='red')
axes[1,1].set_title('Days Absent by Dropout Status')
axes[1,1].legend()

plt.tight_layout()
plt.savefig('../models/feature_distributions.png', dpi=300)
plt.show()
```

```python
# Correlation heatmap
fig, ax = plt.subplots(figsize=(12, 10))
numeric_cols = ['age', 'previous_grade', 'attendance_rate', 'assignments_completed', 
                'test_scores_avg', 'days_absent_last_term', 'late_arrivals', 
                'disciplinary_incidents', 'parent_education_level', 'family_income_level']

corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', ax=ax)
ax.set_title('Correlation Matrix', fontsize=16)
plt.tight_layout()
plt.savefig('../models/correlation_matrix.png', dpi=300)
plt.show()
```

```python
# Interactive scatter plot
fig = px.scatter(
    df, 
    x='attendance_rate', 
    y='test_scores_avg',
    color='Class',
    size='assignments_completed',
    hover_data=['age', 'days_absent_last_term'],
    title='Student Data Visualization',
    labels={'attendance_rate': 'Attendance Rate (%)', 'test_scores_avg': 'Test Score Average (%)'},
    width=800,
    height=600
)
fig.write_html('../models/interactive_scatter.html')
fig.show()

# 3D Scatter Plot
fig_3d = px.scatter_3d(
    df,
    x='attendance_rate',
    y='assignments_completed',
    z='test_scores_avg',
    color='Class',
    title='3D Visualization of Student Features',
    labels={'attendance_rate': 'Attendance', 'assignments_completed': 'Assignments', 'test_scores_avg': 'Test Scores'},
    width=800,
    height=700
)
fig_3d.write_html('../models/interactive_3d.html')
fig_3d.show()
```
